## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains 30,000 rows and 44 columns — one row per article.
Key engagement columns include clicks_90d (mean: 16, max: 4,178), engagement_rate
(mean: 2.5%), and ctr (mean: 0.51%). The clicks_90d distribution is heavily
skewed — the median is just 1 click while the max is 4,178 — a classic long-tail

exactly what makes the problem worth modeling.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('/content/flyrank-internship-ml/data/raw/content_refresh_anonymized.csv')

print(f"Shape: {df.shape}")
print(f"Total articles: {len(df)}")
print(f"\nEngagement column stats:")
print(f"clicks_90d    — mean: {df['clicks_90d'].mean():.1f}, median: {df['clicks_90d'].median():.0f}, max: {df['clicks_90d'].max():.0f}")
print(f"engagement_rate — mean: {df['engagement_rate'].mean():.2f}%, max: {df['engagement_rate'].max():.1f}%")
print(f"ctr           — mean: {df['ctr'].mean():.2f}%, median: {df['ctr'].median():.2f}%")
print(f"\nLong-tail check: 75% of articles have <= {df['clicks_90d'].quantile(0.75):.0f} clicks")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

What this work CAN say:
- "Articles with higher word_count and lower avg_position are associated with
  more clicks_90d in this dataset" (observed, directional)
- "This engagement score can help editors prioritize which articles to review
  first" (decision-support)
- "The model identifies patterns in ctr, engagement_rate, and scroll_rate that
  correlate with clicks in this corpus" (measured association)

What this work CANNOT say:
- "Increasing word count will cause more clicks" (no causal proof)
- "This predicts future Google rankings or impressions" (out of scope)
- "This generalizes to other websites or industries" (this dataset only)
- "trend_pct tells us what will happen next month" (directional signal only)

```python
# Scope check: what signals exist vs what we can claim
engagement_cols = ['clicks_90d', 'engagement_rate', 'ctr', 'scroll_rate', 'pageviews_90d']
feature_cols = ['word_count', 'avg_position', 'content_age_days', 'search_volume', 'days_since_last_update']

print("Observable engagement signals we can USE as targets:")
for col in engagement_cols:
    print(f"  {col}: mean={df[col].mean():.2f}, nulls={df[col].isna().sum()}")

print("\nContent features we can USE as predictors:")
for col in feature_cols:
    print(f"  {col}: mean={df[col].mean():.2f}, nulls={df[col].isna().sum()}")

print("\nClaim boundary: association only, not causation.")
```
